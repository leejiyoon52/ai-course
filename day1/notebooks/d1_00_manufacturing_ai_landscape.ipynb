## 1. AI 생태계 비즈니스 구도 — Enablers와 Adopters

> AI 생태계에는 두 종류의 회사가 있습니다.
> **AI를 만드는 회사**와, **AI로 물건을 더 잘 만드는 회사**입니다.

| | AI Enablers | AI Adopters |
|---|---|---|
| 누구 | NVIDIA, 클라우드 사업자, 파운데이션 모델 개발사 | **제조 기업** (우리가 서 있는 곳) |
| 미션 | 더 큰 모델, 더 빠른 칩 | **물리적 P&L 개선** |
| 투자 | 대규모 CAPEX(설비 투자) — GPU·데이터센터 | 기존 설비·데이터에서 가치 추출 |
| 성공 지표 | 모델 성능, 시장 점유율 | "수율 0.1% 향상", "에너지 비용 5% 절감" |

Adopter의 AI는 데모가 아니라 **손익계산서(P&L)의 숫자**로 평가받습니다.
수율(양품 비율) 0.1%가 반도체 공장에서는 연간 수백억 원입니다.
이 관점이 오늘 하루 모든 기술 선택의 기준이 됩니다.

## 2. MPRS 가치사슬로 보는 제조 AI 스펙트럼

제조 기업의 가치사슬을 M·P·R·S 네 영역으로 나누면 AI 적용 지점이 한눈에 들어옵니다.

- **M (Marketing/SCM)** — 원자재 수요·공급 예측, 가격 최적화
- **P (Production)** ← **오늘의 딥다이브 타겟** — OT 데이터 기반 예지보전(PdM), 공정 최적화(APC), 비전 불량 검사
- **R (R&D)** — 신소재·배터리 분자 시뮬레이션(AI for Science), 특허·실험 이력 RAG
- **S (Support)** — 영상 기반 EHS 위험 감지, 공급망 리스크 모니터링

> **용어 한 줄**
> **OT(Operational Technology)**: 설비·센서·제어기(PLC) 등 공장 운영 기술. IT와 대비되는 개념입니다.
> **PdM(Predictive Maintenance, 예지보전)**: 고장 신호를 미리 읽어 계획 정비로 전환하는 활동입니다.
> **APC(Advanced Process Control, 고급 공정 제어)**: 센서 피드백으로 공정 조건을 실시간 보정하는 제어입니다.

> **현장 노트**
> 모델 이야기 전에 데이터 수집 인프라부터 확인해야 합니다.
> 센서 신호는 PLC(설비 제어기) → SCADA(감시 제어) → MES(생산 관리) 층을 거쳐 쌓입니다.
> 이 배관이 없는 공장에서는 어떤 모델도 시작할 수 없습니다.
> PoC 제안서를 받으면 저는 모델 구조보다 "데이터가 어디에 어떤 주기로 쌓이는가"를 먼저 봅니다.

- **임계값(Threshold)** 은 *알람 민감도 다이얼*입니다 — 돌릴수록 오탐과 미검출이 시소를 탑니다.

> **현장 노트**
> 오탐(False Alarm) 1건의 비용은 라인 정지 비용입니다. 미검출(Miss) 1건의 비용은
> 고객 클레임과 리콜입니다. 두 비용은 대칭이 아니며, 공정마다 비율이 다릅니다.
> 그래서 임계값은 데이터 과학이 아니라 **비용 구조**가 정하는 값에 가깝습니다.
> 이 질문을 던지지 않은 이상탐지 PoC는 현장에서 반드시 되돌아옵니다.

### 6-2. 도메인 시프트 미리보기

> 같은 기종의 설비 두 대. 데이터도 같을까요?

NB02에서 본격적으로 다룰 **항공기 엔진 데이터(C-MAPSS)** 를 잠깐 미리 봅니다.
FD001과 FD003은 **같은 엔진**을 **다른 운전 조건**에서 기록한 데이터셋입니다.

````python
# 같은 엔진, 다른 운전 조건의 두 데이터셋 로드 (FD001 vs FD003)
"""따라하기"""
````

````python
# 같은 센서(s2, s11)의 분포를 겹쳐 그려 본다
fig, axes = plt.subplots(1, 2, figsize=(12, 3.4))
for ax, s in zip(axes, ["s2", "s11"]):
    ax.hist(train_a[s], bins=60, alpha=0.6, label="FD001 (A)", density=True)
    ax.hist(train_b[s], bins=60, alpha=0.6, label="FD003 (B)", density=True)
    ax.set_title(f"sensor {s} distribution")
    ax.legend()
plt.tight_layout()
plt.show()
print("같은 기종·같은 센서인데 분포의 중심이 어긋나 있습니다.")
````

````python
# 눈이 아니라 숫자로 — 두 설비의 센서 평균이 얼마나 어긋났는지 표준편차 단위로 계산
shift = pd.DataFrame({
    "mean_A": train_a[["s2", "s3", "s11", "s15"]].mean(),
    "mean_B": train_b[["s2", "s3", "s11", "s15"]].mean(),
    "std_A": train_a[["s2", "s3", "s11", "s15"]].std(),
})
shift["shift(σ)"] = ((shift["mean_B"] - shift["mean_A"]) / shift["std_A"]).round(2)
print(shift.round(2).to_string())
print("\nshift(σ)가 1을 넘는 센서 — A 기준으로는 '항상 이상'처럼 보이는 값입니다.")
````

**이것이 난제 ③ Concept Drift & Sensor Shift 입니다**

- A 설비 데이터로 학습한 모델을 B 설비에 그대로 얹으면, 모델이 보는 세상 자체가 달라집니다.
- 부품 교체·공정 조건 변경·계절 변화도 정확히 같은 구조의 문제를 만듭니다.
- **NB04에서 이 두 데이터셋으로 성능이 실제로 무너지는 것을 확인하고, 도메인 적응(MMD)으로 복구합니다.**

> **현장 노트**
> 논문 벤치마크에서 F1 0.95를 찍은 모델이 현장에서 무너지는 가장 흔한 이유는
> 설비마다 센서 설치 위치와 샘플링 주기가 다르기 때문입니다.
> 같은 기종이어도 A동 3호기와 B동 7호기는 사실상 다른 데이터입니다.
